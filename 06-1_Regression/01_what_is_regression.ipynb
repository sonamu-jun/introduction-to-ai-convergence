# What Is Regression?

A delivery app predicts 32 minutes, but the food arrives after 38. A cinema estimates tomorrow's audience, and a creator wonders how many views a new clip will receive. Each prediction asks for a number.

Regression learns a relationship between inputs and a numerical outcome from past examples. Classification chooses a category, such as late or on time; regression predicts a quantity, such as delivery time.

Every delivery, movie, park visit, and video in these notebooks is simulated. The numbers illustrate patterns and mistakes, not records from a real service.


## A prediction follows a pattern

Longer delivery routes often take more time, but two equally long trips can end differently. Traffic, restaurant preparation, and other unrecorded details produce variation.

<img src="https://raw.githubusercontent.com/sonamu-jun/introduction-to-ai-convergence/main/06-1_Regression/assets/01_points_and_prediction.png" alt="Delivery records, a fitted trend, and a prediction for a new route." width="640" style="max-width:100%;height:auto;">

Filled dots are recorded delivery times. The line summarizes their pattern; the orange dot is a prediction for a new route. Fitting chooses the line or curve that keeps prediction errors small across many records.

With squared-error fitting, the model targets the average outcome for similar inputs, called the conditional mean. It does not promise the exact result of one delivery. Other fitting criteria can target a median instead.


## Familiar situations and unfamiliar ones

A model fitted to routes from 1 to 8 km has nearby examples for a 4 km trip. This is interpolation: prediction inside the observed input range. A 12 km trip is extrapolation, beyond that range, where the same trend may no longer hold.

Regression usually leaves some records above or below its curve. Numerical interpolation instead joins the recorded dots exactly, including their irregularities.

A fitted pattern alone does not provide an uncertainty interval. New outcomes test whether that pattern is useful. Forecasting applies this idea to the future: earlier records fit the model, and later records check its predictions.
