# What Is Regression?

A thermometer converts a sensor reading into a temperature. Regression learns this connection from pairs of numbers: an input, such as voltage, and an observed result, such as temperature. The fitted rule predicts the result for a new input.

All measurements in these examples are synthetic. The thermometer, student scores, moving cart, and cooling cup illustrate relationships rather than report real observations.

## From measured points to a prediction

Each blue dot is one measurement. Its horizontal position is the input; its vertical position is the observed result. A fitted line follows the overall pattern. A point on that line gives a prediction, which need not equal a measurement.

Repeated measurements at the same input can differ. If three readings at one voltage are 29, 30, and 31 °C, their average is:

$$\text{average temperature} = \frac{29+30+31}{3}=30\ ^\circ\mathrm{C}.$$

In these squared-error examples, regression aims to estimate the average result at a given input, called the conditional mean. A single reading can still be above or below it. Other choices of fitting cost can target a median instead.

## Inside or outside the measured range

Suppose a sensor has been measured from 0 to 5 V.

- Predicting at 2.5 V is interpolation: the input is inside that range.
- Predicting at 6.5 V is extrapolation: the input is outside that range.

Inside the range, nearby measurements support the prediction. Outside it, the prediction depends on the pattern continuing. A sensor that stops responding above 5 V would break that assumption.

Numerical interpolation has a more specific meaning: it joins measurements exactly. Regression usually leaves some points above or below the fitted line. Forecasting means predicting a future result; with time as the input, a time later than every training observation is an extrapolation.
