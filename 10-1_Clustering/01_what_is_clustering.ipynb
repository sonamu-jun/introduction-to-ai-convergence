# 클러스터링의 개념

## 관측값과 비지도학습

- 도시의 배달 수령 위치를 모으면 주문이 자주 모이는 장소를 찾을 수 있다. 각 주문의 소속 구역이 미리 정해져 있지 않아도 위치 사이의 거리를 비교할 수 있다.
- 클러스터링(clustering)은 관측값(observation)의 유사성(similarity)을 바탕으로 데이터를 군집(cluster)으로 묶는 방법이다. 정답 라벨(label) 없이 데이터의 구조를 찾는 비지도학습(unsupervised learning)에 속한다.
- 이 자료의 수령 위치와 공원 방문 위치는 원리를 설명하기 위한 합성 데이터(synthetic data)이며 실제 서비스의 관측이 아니다.

### 분류와 클러스터링

- 분류(classification)는 입력과 정답 라벨(label)의 관계를 학습하여 미리 정해진 클래스(class)를 예측한다.
- 클러스터링은 정답 클래스를 사용하지 않고 관측값을 묶는다. 군집 번호 0·1·2에는 순서나 의미가 없으며, 구성원을 살펴본 뒤 집단의 특성을 해석한다.

## 특징과 유사성

### 위치와 거리

- 특징(feature)은 위치나 방문 횟수처럼 데이터를 비교하는 데 사용하는 속성이다. 수령 위치의 입력은 기준점에서 동서·남북 방향으로 떨어진 거리이다.

  $\displaystyle \mathbf{x}_i=(x_{i1},x_{i2}).$

- $i$는 관측 번호이고 두 좌표의 단위는 km이다. 이 예제는 직선거리가 가까운 위치를 비슷한 관측으로 취급한다.
- 유클리드 거리(Euclidean distance)는 두 관측 사이의 직선거리이다.

  $\displaystyle d(\mathbf{x}_i,\mathbf{x}_j)=\sqrt{(x_{i1}-x_{j1})^2+(x_{i2}-x_{j2})^2}.$

- 각 좌표의 차이가 커질수록 거리가 커진다. 직선거리는 도로의 연결이나 실제 이동 시간을 반영하지 않는다.

### 단위와 표준화

- 연간 지출액과 방문 횟수처럼 단위가 다른 특징을 그대로 비교하면 수치 범위가 큰 특징이 거리를 지배할 수 있다.
- 표준화(standardization)는 각 특징의 평균(mean)을 0, 표준편차(standard deviation)를 1로 맞추는 방법이다. 특징의 상대적 기여와 군집 결과가 달라지므로 분석 목적에 맞게 선택한다.
- 두 좌표가 같은 km 단위를 쓰는 수령 위치 예제는 원래 거리를 사용한다. 공원 예제는 표준화한 좌표에서 거리를 계산한다.


## 군집 기준과 결과 해석

### 중심과 밀도

- K-means는 미리 정한 $K$개의 중심(centroid)을 기준으로 모든 관측을 묶는다. 각 중심 주변에 모인 집단을 표현하는 데 적합하다.
- DBSCAN은 밀도(density)가 높은 이웃(neighborhood)을 연결한다. 군집 수를 미리 지정하지 않으며 어느 군집에도 속하지 않는 관측은 잡음(noise)으로 남길 수 있다.
- 같은 데이터라도 특징·거리·알고리즘·설정에 따라 묶이는 방식이 달라진다. 곡선 모양의 집단은 중심 하나로 표현하기 어려울 수 있다.

### 해석과 적용 한계

- 군집은 선택한 유사성 기준의 결과이며 현실의 구분을 자동으로 확정하지 않는다. 가까운 수령 위치가 묶여도 배송 구역이나 효율적인 경로가 바로 정해지지는 않는다.
- 정답 라벨이 없으면 분류 정확도(accuracy)처럼 맞힌 비율을 바로 계산할 수 없다. 집단 내부의 거리, 집단 간 분리, 데이터나 설정 변화에 대한 안정성을 함께 살펴본다.
- 실제 적용에는 도로·수요·운영 제약 같은 정보가 추가로 필요하다. 잡음으로 분류된 관측도 오류라고 단정할 수 없다.
