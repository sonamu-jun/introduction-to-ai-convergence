# Genetic algorithm (GA)

- A genetic algorithm evolves a population of candidate solutions through selection, crossover, and mutation.
- It evaluates candidate fitness without requiring derivatives.


## Chromosomes and fitness

- A chromosome encodes one candidate solution. In the swing example, its genes specify push directions over time.
- Fitness measures candidate performance. Here, a greater peak swing height means higher fitness.

## Selection, crossover, and mutation

> Evaluate population → select parents → crossover and mutation → evaluate next generation

- Selection: choose fitter parents from small random groups of candidates.
- Crossover: combine the first part of one parent with the remaining part of another to form an offspring.
- Mutation: randomly change individual genes. Here, a mutation reverses a push direction.
- Elitism: copy the best candidate unchanged into the next generation.

## Progress and limitations

- With elitism and a fixed evaluation criterion, the best fitness cannot decrease.
- An offspring can be worse than its parents. Average population fitness can also decrease.
- Loss of diversity can limit exploration, and the global optimum is not guaranteed.
