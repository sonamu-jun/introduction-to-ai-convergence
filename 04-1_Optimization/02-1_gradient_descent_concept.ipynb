# Gradient descent (GD)

- Gradient descent updates decision variables in the direction opposite to the gradient to reduce the objective value.
- The gradient is computed using derivatives.


## Gradient and update direction

<img src="https://raw.githubusercontent.com/sonamu-jun/introduction-to-ai-convergence/main/04-1_Optimization/assets/02_gd_update.png" alt="Gradient descent updates launch speed to reduce landing cost" width="800" style="max-width:100%;height:auto;">

- For one variable, the gradient is the local rate of change of the objective with respect to that variable.
- A positive gradient decreases the variable in the next update; a negative gradient increases it.

> Next value = current value − step size × gradient

- The step size scales the gradient in each update. The actual change depends on both the step size and the gradient.
- For multiple variables, the gradient collects the partial derivatives for all variables.

## Launch speed optimization

- Increase the launch speed when the ball lands short of the target; decrease it when the ball overshoots.
- Evaluate the updated speed and repeat the update.

## Step size and limitations

- A very small step size makes progress slow.
- A very large step size can overshoot the minimum and increase the cost.
- The result depends on the starting point and the objective. Stopping does not generally guarantee a global minimum.
