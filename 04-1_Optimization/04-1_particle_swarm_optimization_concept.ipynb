# Particle swarm optimization (PSO)

- Particle swarm optimization searches with multiple candidates that use their own experience and the best result found by the swarm.
- It evaluates candidates in a continuous search space without requiring derivatives.


## Personal best and global best

<img src="https://raw.githubusercontent.com/sonamu-jun/introduction-to-ai-convergence/main/04-1_Optimization/assets/04_pso_update.png" alt="A particle update using inertia, personal best, and global best" width="640" style="max-width:100%;height:auto;">

- A particle represents one candidate solution. Its position here is a pair of drone controller gains.
- The personal best, pbest, is the best position that a particle has visited.
- The global best, gbest, is the best position found by any particle in the swarm.

## Particle updates

- Inertia carries the previous velocity forward. The cognitive term attracts the particle toward pbest; the social term attracts it toward gbest.

> Next velocity = inertia term + cognitive term + social term

- Weights control the three contributions. Random factors vary the cognitive and social terms to encourage exploration.
- Update the particle position using its velocity, subject to velocity and position bounds.
- Particle position and velocity describe movement in the search space, not the physical motion of the drone.

## Best positions and limitations

- Evaluate each new position and update pbest and gbest when a better result is found.
- The best recorded cost cannot increase, but the average cost of current particles can increase.
- A compact swarm does not guarantee that it has found the global optimum.
