# What Is Classification?

A fruit sorter reports an apple's weight as 180 grams or identifies it as an apple. Regression predicts a numerical quantity. Classification predicts a label from a predefined set of classes.

In supervised classification, each training example pairs input features with a known class label. A feature is a measurable property, such as color or shape. A label records the fruit's identity; it is not a measurement of its size.

The fruit and message datasets in these notebooks are synthetic. The class groups are deliberately distinct enough to make classification visible, with some overlap. Their results do not establish performance on real fruit or messages.


## From measurements to a class

A sorter measures hue, a color angle, and the ratio of length to width. Red is near 0 degrees, orange near 30, and yellow near 60 in the restricted hue range used here. A round fruit has a ratio near one; an elongated fruit has a larger ratio.

The input and possible output labels are written as

$$
\mathbf{x}=(x_1,x_2), \qquad y\in\{\mathrm{Apple},\mathrm{Orange},\mathrm{Banana}\}.
$$

Here, $x_1$ is hue in degrees, $x_2$ is length divided by width, and $y$ is the known class. A classifier estimates a probability $p_k$ for each class $k$ and predicts the class with the largest value:

$$
p_k=P(y=k\mid\mathbf{x}), \qquad \hat y=\underset{k}{\arg\max}\;p_k.
$$

The symbol $\hat y$ denotes the predicted label. The operator $\arg\max$ returns the class with the largest probability, rather than that probability itself.

Binary classification has two possible labels, such as legitimate or spam. Multiclass classification chooses one label, such as apple, orange, or banana. Multilabel classification can assign several tags to the same item.


## Learning a boundary and checking it

A decision boundary separates inputs that a fitted model assigns to different classes. The background regions below come from the same softmax model used in the fruit example. Colored points are training fruit; the star is a new fruit.

Training data determine model parameters. Validation data help choose settings, such as tree depth or a probability threshold. An untouched test set evaluates the final choice. Repeated selection on validation data can also overfit those results.

Overlapping features and missing measurements can cause mistakes. A model can be confident and wrong, especially outside the training range. Calibration checks whether predicted probabilities agree with observed class frequencies.
