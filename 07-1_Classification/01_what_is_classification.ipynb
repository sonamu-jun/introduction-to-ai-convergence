# 분류의 개념

## 지도학습과 예측 대상

- 과일의 색과 크기로 무게를 예측하려면 측정값과 실제 무게를 함께 기록한 사례가 필요하다.
- 지도학습(supervised learning)은 입력과 정답이 짝지어진 데이터에서 관계를 학습하는 방법이다. 학습한 모델은 새로운 입력의 결과를 예측한다.

### 회귀와 분류

- 회귀(regression)는 과일의 무게처럼 수량을 예측한다.
- 분류(classification)는 사과·오렌지처럼 구별되는 범주를 예측한다. 분류의 각 범주를 클래스(class)라고 한다.

### 분류 문제 유형

- 이진 분류(binary classification)는 정상·스팸처럼 두 클래스 중 하나를 고른다.
- 다중 클래스 분류(multiclass classification)는 사과·오렌지·바나나처럼 세 개 이상의 클래스 중 하나를 고른다.

## 측정값–클래스 대응

### 입력 특징과 정답 라벨

- 특징(feature)은 색이나 모양처럼 입력으로 사용하는 속성이다. 분류에서 라벨(label)은 관측한 대상의 실제 클래스를 나타낸다.
- 과일의 색상과 길이/너비를 입력 $\mathbf{x}$로, 실제 종류를 정답 라벨 $y$로 둔다.

  $\displaystyle \mathbf{x}=(x_1,x_2), \qquad y\in\{\text{Apple},\text{Orange},\text{Banana}\}.$

- $x_1$은 색을 각도로 나타내는 색상(hue)이다. 빨강·주황·노랑은 각각 0·30·60도 부근이다.
- $x_2$는 길이/너비로, 둥근 과일에서 1에 가깝고 길쭉한 과일에서 더 크다.

### 학습과 예측

- 학습에서는 특징과 정답 라벨을 함께 사용해 모델의 매개변수(parameter)를 정한다.
- 예측에서는 학습한 모델에 새 과일의 특징을 입력해 예측 라벨 $\hat y$를 구한다. 입력이 달라져도 학습한 매개변수는 그대로이다.

### 클래스 확률과 예측 라벨

- 확률을 출력하는 분류기는 입력 $\mathbf{x}$에 대한 클래스 $k$의 확률 $p_k$를 추정한다. 가장 큰 확률의 클래스를 고르면 다음과 같다.

  $\displaystyle p_k=P(y=k\mid\mathbf{x}), \qquad \hat y=\underset{k}{\arg\max}\;p_k.$

- $\arg\max$는 최댓값 자체가 아니라 그 값을 가진 클래스를 반환한다. 예측 라벨 $\hat y$와 실제 라벨 $y$를 비교하면 분류가 맞았는지 알 수 있다.

### 일반화 성능

- 일반화(generalization)는 학습하지 않은 대상에도 학습한 관계가 적용되는 것이다. 학습 과일을 잘 구분하는 것만으로 새 과일의 성능을 판단할 수는 없다.
- 특징이 겹치거나 학습 범위를 벗어나면 높은 확률의 예측도 틀릴 수 있다. 따라서 별도 자료에서 예측 라벨과 실제 라벨을 비교한다.

### 학습·검증·시험 데이터

- 학습 데이터(training data)는 모델의 매개변수를 정하는 데 사용한다.
- 검증 데이터(validation data)는 트리의 분할 수나 확률 임곗값(threshold)을 선택하는 데 사용한다. 반복해서 선택하면 검증 자료에만 잘 맞는 과적합(overfitting)이 생길 수 있다.
- 시험 데이터(test data)는 학습과 설정 선택에서 제외하고, 최종 모델의 일반화 성능을 평가하는 데 사용한다.
- 이 자료의 과일과 메시지는 합성 데이터(synthetic data)이다. 실제 적용 성능을 판단하려면 해당 환경을 대표하는 시험 자료가 필요하다.
