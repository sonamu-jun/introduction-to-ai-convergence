# What Is Classification?

A delivery app can predict 38 minutes or decide whether an order will be late. The first answer is a quantity; the second is a category. Regression predicts a numerical outcome. Classification assigns an input to a predefined class.

Supervised learning uses examples paired with known labels. A feature describes the input, such as a restaurant's preparation wait. A label records the outcome, such as on time or late. Fitting learns a rule from these pairs; prediction applies that rule to another input.

Every delivery, drink, park visit, and message in these notebooks is simulated. These examples illustrate classification, not the performance of a real service.


## From a feature to a class

A restaurant has kept an order waiting for 18 minutes. A classifier uses that feature to estimate the probability of arriving after a fixed 40-minute promise. It then compares the probability with a decision threshold to choose a class.

Binary classification chooses between two classes. Here, late is the positive class and on time is the negative class. Positive identifies the event of interest; it does not mean a desirable outcome.

Multiclass classification chooses one of several classes, such as tea, coffee, or cocoa. The class names do not imply an order. Assigning several tags to one item is a different task, called multilabel classification.


## Learning a rule and checking it

Training data fit model parameters. Validation data help choose settings, such as tree depth or a decision threshold. An untouched test set evaluates the final choice. Using the same validation results repeatedly can also overfit those results.

A decision boundary separates inputs assigned to different classes. Nearby inputs can receive different labels near this boundary. Missing features and overlapping classes can still cause mistakes on either side.

A probability is an estimate, not a guarantee for an individual case. A confident prediction can be wrong, especially for inputs unlike the training examples. Checking whether predicted probabilities match observed frequencies is called calibration.
